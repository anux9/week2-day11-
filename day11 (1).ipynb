##Task3: التنظيف بقرارات مبرّرة

```python
# الاعمده التصنيفية

print("Gender:", df["Gender"].mode()[0])
print("Married:", df["Married"].mode()[0])
print("Self_Employed:", df["Self_Employed"].mode()[0])
```

```python
# تعبئة القيم الناقصة

df["Gender"] = df["Gender"].fillna("Male")
df["Married"] = df["Married"].fillna("Yes")
df["Self_Employed"] = df["Self_Employed"].fillna("No")

print(df[["Gender", "Married", "Self_Employed"]].isna().sum())
```

```python
# تعديل Dependents

print(df["Dependents"].mode()[0])
```

```python
# تعبئة القيم الناقصة
df["Dependents"] = df["Dependents"].fillna("0")

#تحويل 3+ الى 3
df["Dependents"] = df["Dependents"].replace("3+", "3")

#تحويل العمود الى ارقام
df["Dependents"] = df["Dependents"].astype(int)

print(df["Dependents"].value_counts())
print(df["Dependents"].dtype)
```

```python
# LoanAmount

print("Mean:", df["LoanAmount"].mean())
print("Median:", df["LoanAmount"].median())
```

```python
# نستخدم الوسيط 128 لتعبئة القيم الناقصة لأنه أقل تأثرًا بالقيم المرتفعة

df["LoanAmount"] = df["LoanAmount"].fillna(df["LoanAmount"].median())
```

```python
print(df["LoanAmount"].isna().sum())
```

```python
# Loan_Amount_Term

print(df["Loan_Amount_Term"].mode()[0])
```

أكثر قيمة متكررة هي 360 شهرًا

نحولها إلى سنوات:

360 ÷ 12 = 30 سنة

```python
# التعبئة بـ 360

df["Loan_Amount_Term"] = df["Loan_Amount_Term"].fillna(360)

print(df["Loan_Amount_Term"].isna().sum())
```

```python
# تنظيف Credit_History

df["Credit_History"] = df["Credit_History"].fillna(-1)

print(df.groupby("Credit_History")["Loan_Status"].value_counts(normalize=True) * 100)
```

نلاحظ أن نسبة الموافقة للأشخاص اللي بياناتهم ناقصة
(74%) قريبة من أصحاب السجل الائتماني 1 (79.58%)، وبعيدة عن أصحاب السجل 0 (7.87%).


بناءً على النتائج، نقدر نعوّض القيم الناقصة بالرقم 1 لأن نسبة الموافقة في مجموعة القيم الناقصة أقرب لمجموعة 1

```python
df["Credit_History"] = df["Credit_History"].replace(-1, 1)
print(df["Credit_History"].isna().sum())
print(df["Credit_History"].value_counts())
```

```python
# حساب outliers باستخدام IQR

columns = ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, ":", len(outliers))
```

ملاحظة:


مو معناته إن هذي القيم أخطاءممكن تكون حالات حقيقية، مثل شخص دخله
مرتفع أو طلب قرض بمبلغ كبير.

## Task5: سجلات متحف الديناصورات

```python
import pandas as pd

fossils = {
    "fossil_id": ["F01", "F02", "F03", "F04", "F05", "F02", "F06", "F07", "F08", "F09"],
    "species": ["T-Rex", "Triceratops", "t-rex", "Velociraptor", "Stegosaurus", "Triceratops", "Brachiosaurus", "T-Rex", "Velociraptor", "Stegosaurus"],
    "age_million_years": [68, 67, 66, 75, -150, 67, 150, 68, None, 155],
    "length": ["12 m", "9 m", "40 ft", "2 m", "9 m", "9 m", "26 m", "12.3 m", "6.6 ft", "9 m"],
    "found_in": ["USA", "USA", "Canada", "Mongolia", "USA", "USA", "Tanzania", "USA", "Mongolia", "Portugal"]
}

df = pd.DataFrame(fossils)

print(df)
```

```python
# حذف المكرر وتوحيد اسمائ الانواع

df = df.drop_duplicates()

df["species"] = df["species"].str.title()

print(df)
```

```python
# عندنا عمر سالب -150 , نصححه

df.loc[df["age_million_years"] < 0, "age_million_years"] = 150

print(df)
```

```python
# ملء العمر الناقص

df["age_million_years"] = df["age_million_years"].fillna(
    df.groupby("species")["age_million_years"].transform("median") )

print(df)
```

تعوض العمر الناقص بـ 75 مليون سنة، وهو وسيط الاعمار الموجودة في البيانات

```python
# توحيد وحدات الطول

df[["length_value", "unit"]] = df["length"].str.split(" ", expand=True)

df["length_value"] = df["length_value"].astype(float)

df["length_m"] = df["length_value"]

df.loc[df["unit"] == "ft", "length_m"] = (
    df["length_value"] * 0.3048 )

print(df)
```

```python
# متوسط طول كل نوع، وأقدم نوع، والدولة التي وجد فيها اكثر عدد

print("Average length by species:")
print(df.groupby("species")["length_m"].mean().sort_values())

print("\nOldest species:")
print(df.loc[df["age_million_years"].idxmax(), "species"])

print("\nCountry with most fossils:")
print(df["found_in"].value_counts().idxmax())
```
